```python
import pandas as pd
import unicodedata


maps = {
    'est_socio': {
        1: 'Bajo', 2: 'Medio bajo', 3: 'Medio alto', 4: 'Alto'
    },
    'clase_hog': {
        1: 'Unipersonal', 2: 'Nuclear', 3: 'Ampliado',
        4: 'Compuesto', 5: 'Corresidente'
    },
    'sexo_jefe': {
        1: 'Hombre', 2: 'Mujer'
    },
    'educa_jefe': {
        1: 'Sin instrucción', 2: 'Preescolar', 3: 'Primaria incompleta',
        4: 'Primaria completa', 5: 'Secundaria incompleta',
        6: 'Secundaria completa', 7: 'Preparatoria incompleta',
        8: 'Preparatoria completa', 9: 'Profesional incompleta',
        10: 'Profesional completa', 11: 'Posgrado'
    }
}


vars_to_keep = [
    'folioviv', 'foliohog', 'ubica_geo',
    'est_socio', 'clase_hog', 'sexo_jefe',
    'edad_jefe', 'educa_jefe',
    'tot_integ', 'ocupados', 'ing_cor',
    'jubilacion', 'becas', 'bene_gob', 'gasto_mon'
]


df = pd.read_csv(
    'conjunto_de_datos_concentradohogar_enigh2022_ns.csv',
    usecols=vars_to_keep
)


for col, mapping in maps.items():
    df[col] = df[col].map(mapping).astype('category')


df_map = pd.read_csv('ubicaciongeo.csv', encoding='latin1')


df = df.merge(
    df_map[['ubica_geo', 'desc_ent', 'desc_mun']],
    on='ubica_geo',
    how='left'
)


df = df.drop(columns=['desc_mun', 'folioviv', 'foliohog'], errors='ignore')


final_cols = ['ubica_geo', 'desc_ent'] + [c for c in df.columns if c not in ('ubica_geo','desc_ent')]
df_final = df[final_cols]

df_final.to_csv('concentradohogarbueno.csv', index=False)
print(" Archivo guardado como 'concentradohogarbueno.csv'")
```
