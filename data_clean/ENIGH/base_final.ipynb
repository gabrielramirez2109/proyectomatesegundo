```python
import pandas as pd

df = pd.read_csv('concentradohogarbueno.csv')

numeric_vars = ['tot_integ', 'ocupados', 'ing_cor', 'jubilacion', 'becas', 'bene_gob', 'gasto_mon']
categorical_vars = ['est_socio', 'clase_hog', 'sexo_jefe', 'educa_jefe']

df_num = df.groupby('desc_ent')[numeric_vars].mean().reset_index()
df_cat = df.groupby('desc_ent')[categorical_vars] \
           .agg(lambda x: x.mode().iat[0] if not x.mode().empty else None) \
           .reset_index()

df_summary = pd.merge(df_num, df_cat, on='desc_ent')
df_summary.to_csv('contreadoprom.csv', index=False)
df_summary.head()
```
