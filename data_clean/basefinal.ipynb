```python
import pandas as pd

# 1) Carga de la tabla resumida de ENIGH por estado
df_res = pd.read_csv('contreadoprom.csv')

# 2) Carga y procesamiento de PIB estatal (2022, precios de 2018)
df_pib = pd.read_csv('conjunto_de_datos_piber_actividad_pib2023_p.csv')
mask = df_pib['Descriptores'].str.contains(
    r'Millones de pesos a precios de 2018\|Producto interno bruto, a precios de mercado'
)
df_pib = df_pib[mask].copy()
df_pib['desc_ent'] = df_pib['Descriptores']\
    .str.extract(r'\|Producto interno bruto, a precios de mercado\|(.*?)<')[0]\
    .str.strip()
df_pib = df_pib[['desc_ent','2022']].rename(columns={'2022':'pib_millones'})
df_pib['pib_millones'] = pd.to_numeric(df_pib['pib_millones'], errors='coerce')

# 3) Carga y procesamiento de escolaridad promedio 2022
df_edu = pd.read_csv('Educacion_13.csv', encoding='latin1')
df_edu = df_edu.rename(columns={
    'Entidad federativa':'desc_ent',
    '2022/2023':'escolaridad_promedio'
})
df_edu['desc_ent'] = df_edu['desc_ent'].str.strip()
df_edu['escolaridad_promedio'] = pd.to_numeric(
    df_edu['escolaridad_promedio'], errors='coerce'
)
df_edu = df_edu[['desc_ent','escolaridad_promedio']]

# 4) Merge de las tres fuentes por estado
df_final = (
    df_res
    .merge(df_pib, on='desc_ent', how='left')
    .merge(df_edu, on='desc_ent', how='left')
)

# 5) Guardar el resultado combinado
df_final.to_csv('base_final_combinada.csv', index=False)
print(df_final.head())
```
