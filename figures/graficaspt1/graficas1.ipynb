```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('concentradohogarbueno.csv')

order = [
    'Sin instrucción', 'Preescolar', 'Primaria incompleta',
    'Primaria completa', 'Secundaria incompleta', 'Secundaria completa',
    'Preparatoria incompleta', 'Preparatoria completa',
    'Profesional incompleta', 'Profesional completa', 'Posgrado'
]
df['educa_jefe'] = pd.Categorical(df['educa_jefe'], categories=order, ordered=True)

edu_counts = df['educa_jefe'].value_counts().sort_index()
colors = plt.cm.tab20(range(len(edu_counts)))
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(edu_counts, labels=edu_counts.index, autopct='%1.1f%%', startangle=140, colors=colors)
ax.axis('equal')
plt.savefig('educa_jefe_pie.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('educa_jefe_pie.png')

sex_counts = df['sexo_jefe'].value_counts()
sex_colors = ['skyblue' if s == 'Hombre' else 'lightpink' for s in sex_counts.index]
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(sex_counts.index, sex_counts.values, color=sex_colors)
plt.xticks(rotation=0)
plt.savefig('sexo_jefe_bar.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('sexo_jefe_bar.png')

becas_per_state = df[df['becas'] > 0].groupby('desc_ent').size().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(becas_per_state.index, becas_per_state.values, color='steelblue')
plt.xticks(rotation=90)
plt.savefig('becas_por_estado.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('becas_por_estado.png')

bene_gob_per_state = df[df['bene_gob'] > 0].groupby('desc_ent').size().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(bene_gob_per_state.index, bene_gob_per_state.values, color='seagreen')
plt.xticks(rotation=90)
plt.savefig('bene_gob_por_estado.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('bene_gob_por_estado.png')
```
