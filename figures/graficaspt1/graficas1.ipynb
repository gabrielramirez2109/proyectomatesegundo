```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('concentradohogarbueno.csv')

order = [
    'Sin instrucción', 'Preescolar', 'Primaria incompleta',
    'Primaria completa', 'Secundaria incompleta', 'Secundaria completa',
    'Preparatoria incompleta', 'Preparatoria completa',
    'Profesional incompleta', 'Profesional completa', 'Posgrado'
]
df['educa_jefe'] = pd.Categorical(df['educa_jefe'], categories=order, ordered=True)

# Pie chart: educa_jefe
edu_counts = df['educa_jefe'].value_counts().sort_index()
colors = plt.cm.tab20(range(len(edu_counts)))
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(edu_counts, labels=edu_counts.index, autopct='%1.1f%%', startangle=140, colors=colors)
ax.axis('equal')
plt.title('Nivel de escolaridad del jefe de hogar')
plt.savefig('educa_jefe_pie.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('educa_jefe_pie.png')

# Bar chart: sexo_jefe with percentages
sex_counts = df['sexo_jefe'].value_counts()
total_sex = sex_counts.sum()
fig, ax = plt.subplots(figsize=(6, 4))
bars = ax.bar(sex_counts.index, sex_counts.values, color=['skyblue', 'lightpink'])
for bar, count in zip(bars, sex_counts.values):
    perc = count / total_sex * 100
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{perc:.1f}%', ha='center', va='bottom')
plt.xticks(rotation=0)
plt.title('Distribución del sexo del jefe de hogar')
plt.savefig('sexo_jefe_bar.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('sexo_jefe_bar.png')

# Bar chart: hogares con becas por entidad with percentages
becas_counts = df[df['becas'] > 0].groupby('desc_ent').size().sort_values(ascending=False)
total_becas = becas_counts.sum()
fig, ax = plt.subplots(figsize=(12, 6))
bars = ax.bar(becas_counts.index, becas_counts.values, color='steelblue')
for bar, count in zip(bars, becas_counts.values):
    perc = count / total_becas * 100
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{perc:.1f}%', ha='center', va='bottom')
plt.xticks(rotation=90)
plt.title('Hogares con becas por entidad')
plt.ylabel('Número de hogares')
plt.savefig('becas_por_estado.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('becas_por_estado.png')

# Bar chart: hogares con beneficios gubernamentales por entidad with percentages
bene_counts = df[df['bene_gob'] > 0].groupby('desc_ent').size().sort_values(ascending=False)
total_bene = bene_counts.sum()
fig, ax = plt.subplots(figsize=(12, 6))
bars = ax.bar(bene_counts.index, bene_counts.values, color='seagreen')
for bar, count in zip(bars, bene_counts.values):
    perc = count / total_bene * 100
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{perc:.1f}%', ha='center', va='bottom')
plt.xticks(rotation=90)
plt.title('Hogares con beneficios gubernamentales por entidad')
plt.ylabel('Número de hogares')
plt.savefig('bene_gob_por_estado.png', dpi=300, bbox_inches='tight')
plt.show()
files.download('bene_gob_por_estado.png')

```
